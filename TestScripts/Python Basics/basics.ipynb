# Single line & multiline Comments
Comments - Comments are notes in source code for people reading it. Pythn ignores comments when it runs the program

Single line Comments  -  Start a comment with # or ##. Everything after # or ##  on that line is ignored by python.

Multi-Line comments - Python has no special multiline-comment syntax. For a comment spanning multiple lines, put `#` at the start of each line or we can use Triple-quoted text (`'''...'''` or `"""..."""`) is a string literal. Triple-quoted strings are commonly used for docstrings, which document modules, functions, and classes.

# Defination of Syntax & Semantics
1. Syntax - syntax is the set of rules that defines the combination of symbols are considered to be correctly structured programs in a language. In simple terms, syntax is about the correct arrangements of words and symbols in a code

2. Semantics - semantics refers to the interpretation of symbols, char and commands in a language. It is about what the code is supposed to do when it runs.

## Indentation 
Indentation is used to define the structure and hierarchy of the code. unlikely many other programming languages are using braces {} to delimit the block of code.
pyton uses indentation to determine the grouping of statements. this means all the statements within a block must be indented at the same level.


```python
# Indentation - python uses indentation to define blocks of code. Consistent use of spaces or a tab is required

age=32
if age>18:
    print("Age is greater than 18")
else:
    print("Age is less than 18")
```
