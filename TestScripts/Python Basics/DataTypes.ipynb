```python
#Integer Example - An integer is a number without a decimal point.

age = 35
type(age)

#floating point data Type - A floating-point number is a number that contains a decimal point.
# or In Simple terms, Float = Decimal numbers

height = 1.75
print(height)
print(type(height))

#Strings - A String is a sequence of characters enclosed in a single or double quotes.

name = 'kalyan'
print(name)
print(type(name))

#Boolean - A Boolean is a data type that can have 2 possible values : True or False.
is_student = True
print(is_student)
print(type(is_student))


```
