# Sets in Python

## 1. What is a Set in Python?

A set in Python is a collection of unique elements. It does not allow duplicate values, and it is unordered, meaning the elements do not have fixed positions.

Sets are useful when we want to store unique values, remove duplicates, or perform mathematical set operations.

## 2. Characteristics of Sets

- **Unique:** A set does not allow duplicate elements.
- **Unordered:** Elements do not have a guaranteed order.
- **Mutable:** We can add or remove elements after creating a set.
- **No indexing:** We cannot access elements using positive or negative indexes.
- **Different data types:** A set can contain different types of hashable elements.
- **No unhashable elements:** Lists, dictionaries, and sets cannot be directly stored as elements.
- **Dynamic size:** The size of a set can increase or decrease when elements are added or removed.

## 3. Creating a Set

Sets can be created using curly braces or the built-in `set()` function.

To create an empty set, we must use `set()` because empty curly braces create a dictionary.

## 4. Adding Elements to a Set

- `add()` — adds a single element.
- `update()` — adds multiple elements from an iterable.

Adding an element that already exists does not create a duplicate.

## 5. Removing Elements from a Set

- `remove()` — removes an element and raises a `KeyError` if it does not exist.
- `discard()` — removes an element without raising an error if it does not exist.
- `pop()` — removes and returns an arbitrary element.
- `clear()` — removes all elements from the set.

## 6. Set Operations

- **Union:** Combines all unique elements from two sets.
- **Intersection:** Returns the elements common to both sets.
- **Difference:** Returns elements present in one set but not the other.
- **Symmetric Difference:** Returns elements present in either set, but not in both.

## 7. Membership Testing

- `in` — checks whether an element exists in a set.
- `not in` — checks whether an element does not exist in a set.

Membership testing is generally very efficient for sets.

## 8. Set Comparison Methods

- `issubset()` — checks whether all elements of one set belong to another.
- `issuperset()` — checks whether a set contains all elements of another set.
- `isdisjoint()` — checks whether two sets have no common elements.

## 9. Iterating Through a Set

We can use a `for` loop to access elements one by one.

However, because sets are unordered, we should not depend on a particular iteration order.

## 10. Advantages of Sets

- Automatically eliminate duplicate values.
- Provide efficient membership testing.
- Support mathematical set operations.
- Help identify common and unique values.

## 11. Limitations of Sets

- They do not support indexing or slicing.
- They do not guarantee element order.
- They cannot contain mutable, unhashable elements such as lists and dictionaries.
- They are not suitable when duplicate values or positional access are required.

## 12. Interview Definition
