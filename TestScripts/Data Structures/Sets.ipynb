**Sets in Python**

Sets are built in data type in python used to store collection of unique items. They do not follow a specific order, do not allow duplicates elements.

sets are useful for membership tests, eliminaing duplicate entries and performing mathematical set operations like union, intersection, difference and symmetric difference.

```python
## Create a Set
my_set ={1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```
