## 1. Introduction to Tuples

**Definition:** A tuple in Python is an ordered collection of elements that is immutable, meaning its elements cannot be modified directly after the tuple is created.

### Key Characteristics of Tuples

- **Ordered:** Elements maintain their original positions.
- **Immutable:** Elements cannot be reassigned, added, or removed directly after creation.
- **Allows duplicates:** The same value can appear multiple times.
- **Supports multiple data types:** A tuple can contain integers, strings, floats, Booleans, and other objects.
- **Uses parentheses:** Tuples are commonly created using `()`.
- **Supports indexing and slicing:** Elements can be accessed by position or extracted as a range.

Example:

```python
my_tuple = (10, 20, 30, "Python", True)
print(my_tuple)
```

Output:

```text
(10, 20, 30, 'Python', True)
```

## 2. Creating Tuples

**Definition:** Creating a tuple means grouping multiple elements into a single collection.

### A. Creating a tuple with parentheses

```python
numbers = (10, 20, 30, 40)
print(numbers)
```

### B. Creating a tuple without parentheses

```python
numbers = 10, 20, 30
print(numbers)
```

Python creates a tuple from the comma-separated values.

### C. Creating a single-element tuple

```python
single = (10,)
print(type(single))
```

Output:

```text
<class 'tuple'>
```

**Note:** The comma is essential. `(10)` is an integer, whereas `(10,)` is a tuple.

### D. Creating an empty tuple

```python
empty_tuple = ()
print(empty_tuple)
```

Output:

```text
()
```

## 3. Accessing Tuple Elements

**Definition:** Accessing a tuple means retrieving its elements using indexing or slicing.

### A. Positive Indexing

Indexing starts at `0`, which represents the first element.

```python
fruits = ("Apple", "Banana", "Mango")
print(fruits[0])
```

Output:

```text
Apple
```

### B. Negative Indexing

Negative indexing starts from the end of the tuple. The index `-1` represents the last element.

```python
fruits = ("Apple", "Banana", "Mango")
print(fruits[-1])
```

Output:

```text
Mango
```

### C. Tuple Slicing

Slicing extracts a portion of a tuple using the syntax `[start:stop:step]`.

```python
numbers = (10, 20, 30, 40, 50)
print(numbers[1:4])
```

Output:

```text
(20, 30, 40)
```

The stop index is excluded from the result.

## 4. Tuple Operations

**Definition:** Tuple operations allow you to combine tuples, repeat their elements, check membership, and determine their length.

| Operation | Description | Example |
|---|---|---|
| Concatenation (`+`) | Combines two tuples. | `(1, 2) + (3, 4)` |
| Repetition (`*`) | Repeats tuple elements. | `(1, 2) * 2` |
| Membership (`in`) | Checks whether an element exists. | `2 in (1, 2, 3)` |
| Non-membership (`not in`) | Checks whether an element is absent. | `5 not in (1, 2, 3)` |
| Length (`len()`) | Returns the number of elements. | `len((1, 2, 3))` |

Example:

```python
a = (1, 2)
b = (3, 4)

print(a + b)
print(a * 2)
print(2 in a)
print(len(b))
```

Output:

```text
(1, 2, 3, 4)
(1, 2, 1, 2)
True
2
```

## 5. Understanding Tuple Immutability

**Definition:** Immutability means that the elements of a tuple cannot be changed, added, or removed directly after the tuple has been created.

Example:

```python
numbers = (10, 20, 30)

numbers[1] = 50  # Raises TypeError
```

Output:

```text
TypeError: 'tuple' object does not support item assignment
```

Python raises an error because tuple elements cannot be reassigned.

Although you cannot modify the original tuple, you can create a new tuple and assign it to the same variable.

```python
numbers = (10, 20, 30)
numbers = numbers + (40,)

print(numbers)
```

Output:

```text
(10, 20, 30, 40)
```

**Important:** A tuple cannot be modified directly, but if it contains a mutable object such as a list, that inner object may still be changed.

## 6. Tuple Methods

**Definition:** Tuple methods are built-in methods used to count occurrences of an element and locate the first occurrence of a value.

Python tuples provide two built-in methods:

| Method | Purpose | Example |
|---|---|---|
| `count()` | Returns the number of times a value appears. | `t.count(10)` |
| `index()` | Returns the index of the first occurrence of a value. | `t.index(20)` |

### A. `count()`

```python
numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))
```

Output:

```text
3
```

### B. `index()`

```python
numbers = (10, 20, 30, 20)
print(numbers.index(20))
```

Output:

```text
1
```

The result is `1` because the first occurrence of `20` is at index `1`.

**Note:** If the specified value does not exist, `index()` raises a `ValueError`.

## 7. Tuple Packing and Unpacking

### A. Tuple Packing

**Definition:** Tuple packing is the process of combining multiple values into a single tuple. Python creates the tuple from comma-separated values.

```python
student = "Kalyan", 25, "Python"
print(student)
```

Output:

```text
('Kalyan', 25, 'Python')
```

### B. Tuple Unpacking

**Definition:** Tuple unpacking assigns individual tuple elements to separate variables.

```python
student = ("Kalyan", 25, "Python")

name, age, course = student

print(name)
print(age)
print(course)
```

Output:

```text
Kalyan
25
Python
```

For ordinary unpacking, the number of variables must match the number of elements in the tuple.

### C. Swapping Values Using Unpacking

Python allows you to swap two variables without using a temporary variable.

```python
a = 10
b = 20

a, b = b, a

print(a, b)
```

Output:

```text
20 10
```

## 8. Nested Tuples

**Definition:** A nested tuple is a tuple that contains one or more tuples as its elements.

Example:

```python
students = (
    ("Kalyan", 25),
    ("Ravi", 22),
    ("Priya", 24)
)

print(students)
```

Output:

```text
(('Kalyan', 25), ('Ravi', 22), ('Priya', 24))
```

### Accessing Nested Tuple Elements

```python
print(students[0])
print(students[0][0])
print(students[0][1])
```

Output:

```text
('Kalyan', 25)
Kalyan
25
```

- `students[0]` accesses the first inner tuple.
- `students[0][0]` retrieves the student's name.
- `students[0][1]` retrieves the student's age.

## List vs Tuple – Interview Answer

| Difference         | List                                                                     | Tuple                                                                       |
| ------------------ | ------------------------------------------------------------------------ | --------------------------------------------------------------------------- |
| Definition         | A list is a collection of elements that can be modified after creation.  | A tuple is a collection of elements that cannot be modified after creation. |
| Syntax             | Lists are created using square brackets `[]`.                            | Tuples are generally created using parentheses `()`.                        |
| Mutability         | A list is mutable, so we can add, remove, or modify elements.            | A tuple is immutable, so we cannot add, remove, or modify its elements.     |
| Example            | `numbers = [10, 20, 30]`                                                 | `numbers = (10, 20, 30)`                                                    |
| Adding elements    | We can add elements using `append()` or `insert()`.                      | We cannot directly add elements to an existing tuple.                       |
| Removing elements  | We can remove elements using `remove()` or `pop()`.                      | We cannot directly remove elements from an existing tuple.                  |
| Modifying elements | We can modify elements using their index.                                | We cannot modify elements using their index.                                |
| Duplicate values   | A list allows duplicate values.                                          | A tuple also allows duplicate values.                                       |
| Data types         | A list can store different data types.                                   | A tuple can also store different data types.                                |
| Performance        | Lists may have slightly more overhead because they support modification. | Tuples can be slightly more memory-efficient for similar data.              |
| Use case           | Lists are useful when data changes frequently.                           | Tuples are useful when data should remain unchanged.                        |

### Interview conclusion

“The main difference between a list and a tuple is mutability. A list is mutable, whereas a tuple is immutable. Lists use square brackets, while tuples generally use parentheses.”
