**Python Lists – Introduction, Definitions, and Video Outline**

**Introduction to Lists**

Definition: A list in Python is an ordered and mutable collection of items that allows you to store multiple values in a single variable.

**Key characteristics of lists:**
1. Ordered: Items maintain their position in the list.
2. Mutable: Items can be added, removed, or modified after the list is created.
3. Allows different data types: A list can contain integers, strings, floats, Booleans, and other data types.
4. Allows duplicates: The same value can appear multiple times.
5. Uses square brackets: Lists are created using [].

**Video outLine:**
1. **Intro to Lists:** -

2. **Creating lists:** - Creating a list means storing multiple items in a single variable using square brackets [].

3. **Accesing and modifying list elements:** - Accessing means retrieving an element using its index, while modifying means changing an existing element's value.

4. **list methods** - List methods are built-in operations used to add, remove, search, or rearrange items in a list.

Common methods include:
- append() – Adds an item to the end.
- insert() – Adds an item at a specified position.
- remove() – Removes the first matching item.
- pop() – Removes and returns an item.
- sort() – Sorts the list.
- reverse() – Reverses the order of items.
- extend() – Adds items from another iterable.

5. **Slicing lists:** - List slicing is used to extract a portion of a list by specifying start, stop, and step values.

6. **Iterating over Lists** - Iterating over a list means accessing or processing its elements one by one using a loop

7. **list comprehensions** - A list comprehension is a concise way to create a new list by applying an expression to each item in an iterable, optionally filtering items using a condition.

8. **Nested lists** - A nested list is a list that contains one or more lists as its elements.
