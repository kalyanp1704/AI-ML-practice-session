# Round 1: Technical Round

## Level 1: Basics

### Q1. Please introduce yourself technically.

I am a QE Automation Engineer with 5+ years of experience in iGaming,
FinTech and E-Commerce. I work mainly with Playwright and JavaScript,
and earlier I built Selenium with Java and TestNG frameworks. Right now
I am at Ideyalabs on the Fanatics sportsbook and casino platform, where
I built the Playwright framework for the casino application. Along with
UI automation, I do API testing with Postman, SQL validation, and CI/CD
with GitHub Actions and Jenkins. Recently I have also used AI-assisted
QA workflows using GitHub Copilot, Claude Code and Kiro to speed up test
creation and scripting.

### Q2. What is the difference between QA, QC and Testing?

QA is about the process: making sure we follow the right practices so
defects are prevented. QC is about the product: checking that what we
built meets the requirements. Testing is the main activity under QC,
where we actually run the application and find defects. In my role as a
QE, I do a bit of all three: I help define test standards, I write
automation, and I verify the product.

### Q3. What is the difference between SDLC and STLC?

SDLC covers the whole software life cycle: requirement, design,
development, testing, deployment and maintenance. STLC is only the
testing part inside it: requirement analysis, test planning, test case
design, environment setup, execution, and closure. In Agile sprints I
follow STLC in small cycles, so I review stories, prepare test
scenarios, automate, execute and report within the same sprint.

### Q4. What is the difference between smoke, sanity and regression testing?

Smoke testing is a quick check on a new build to see if the main
features work, such as login, lobby loading and game launch. Sanity
testing is a narrow check after a small fix or change to confirm that
the fix works. Regression testing is the wider check to confirm that new
changes did not break existing features. In my projects, regression is
the biggest automated suite and runs through CI.

### Q5. What types of testing have you done?

Functional testing, regression testing, cross-browser testing, API
testing and database validation using SQL. On the Casino project, that
means things like game launch, bet placement, favorites and promotions.
On Coyni, it was FinTech workflows. On Wish, it was registration, cart
and payment validation.

### Q6. Explain the defect life cycle.

A defect starts as New, then gets Assigned to a developer and becomes
Open or In Progress. After the fix it moves to Fixed, and QA retests it.
If it works, it is Verified and Closed. If it still fails, it is
Reopened. A defect can also be Rejected, Duplicate or Deferred depending
on the case. I have managed this life cycle in Jira, and recently I have
used AI to help write clear defect summaries.

### Q7. What is the difference between severity and priority? Give an example.

Severity is how badly the defect affects the system. Priority is how
soon it should be fixed. Example from casino: if a game does not launch
for any user, that is high severity and high priority. A spelling
mistake in the brand name on the home page is low severity but may be
high priority because it is very visible.

### Q8. What is an RTM and why do you use it?

Requirement Traceability Matrix maps each requirement to its test cases
and defects. It shows that every requirement is covered and nothing is
missed. With my spec-driven approach on Fanatics, traceability from
requirement to test became much easier because tests were generated and
organised against the requirement itself.

### Q9. What should you automate and what should you not?

I automate repeatable, stable and high-value tests: regression, smoke,
data-driven flows and critical user journeys like login, game launch and
bet placement. I avoid automating one-time tests, very unstable
features, or tests that need heavy visual judgement. The goal is to save
manual effort and give quick feedback, not to automate everything.

### Q10. What is the Page Object Model and why do you use it?

POM means creating a separate class for each page or component, which
holds its locators and actions. The test only calls these methods. The
big benefit is maintenance: if a locator changes, I fix it in one place
and all tests keep working. I used POM in both Selenium (Java) on Coyni
and Wish, and in Playwright (TypeScript) on Fanatics.

### Q11. What are the common TestNG annotations and what is a Data Provider?

The main ones are `@BeforeSuite`, `@BeforeClass`, `@BeforeMethod`,
`@Test`, `@AfterMethod`, `@AfterClass` and `@AfterSuite`. They control
setup and teardown. A Data Provider lets one test run multiple times
with different data sets. On Wish I used Data Providers for data-driven
suites, for example running registration and payment validation with
different inputs.

### Q12. Which Git commands do you use daily?

`git clone`, `git pull`, `git checkout -b` for a feature branch,
`git add`, `git commit`, `git push`, `git status`, `git diff` and
`git merge`. I raise pull requests for framework changes so teammates
can review before merging. I also use rebase when I want a clean
history, and stash when I need to switch branches quickly.

### Q13. Selenium vs Playwright: what is the difference and why did you move to Playwright?

Selenium drives the browser through WebDriver and needs explicit waits
and more setup. Playwright has built-in auto-waiting, faster execution,
easy parallel runs, tracing, and good support for modern web apps with
multiple tabs and contexts. For Fanatics, which is a dynamic casino and
sportsbook UI, Playwright with TypeScript gave us more stable and faster
tests. I still know Selenium well and have used it on Coyni and Wish.

## Level 2: Intermediate

### Q14. How do you choose locators in Playwright?

I prefer user-facing and stable locators first: `getByRole`,
`getByText`, `getByLabel`, and `data-testid` where the app provides it.
I avoid long absolute XPath and auto-generated class names because they
break often. If the app has no stable attributes, I ask developers to
add test IDs. This keeps tests readable and reduces maintenance.

### Q15. What is auto-waiting in Playwright?

Before an action like click or fill, Playwright automatically waits
until the element is attached, visible, stable and enabled. That removes
most of the manual waits we write in Selenium and reduces flaky tests.
Web-first assertions like `expect(locator).toBeVisible()` also retry
until the condition is met or the timeout ends.

### Q16. What types of waits are there in Selenium? Why avoid Thread.sleep?

Implicit wait, explicit wait (`WebDriverWait` with `ExpectedConditions`)
and fluent wait. I mostly use explicit waits for specific conditions.
`Thread.sleep` always waits the full time even if the element is ready,
so it slows the suite, and it can still fail if the page is slower than
expected.

### Q17. How do you handle frames, new tabs, alerts and dynamic elements?

In Selenium I switch using `driver.switchTo()` for frames, alerts and
windows, and I use explicit waits for dynamic elements. In Playwright,
new tabs and popups are handled through the page and context objects,
frames through `frameLocator`, and dialogs through the dialog event.
This is useful in casino testing because games often launch in a new tab
or inside an iframe.

### Q18. How is your Playwright framework structured?

I keep it layered: tests at the top, then page objects and reusable
components, then helpers and utilities, then test data and
configuration. The Playwright config handles browsers, retries,
timeouts, base URL and parallel workers. I use fixtures to share setup
such as login or preconditions, so tests stay short and readable.
Reports and traces are generated automatically for debugging.

### Q19. Tell me about the Hybrid framework you built on Coyni.

It combined Page Object Model with a data-driven approach, built on
Selenium WebDriver, Java and TestNG. I created reusable keyword
libraries for common actions and a module to manage dependencies and
preconditions, so a test does not have to repeat setup steps. This
reduced code duplication by more than 50% and made the tests more stable
during Agile sprints.

### Q20. How did you run tests in parallel and across browsers on Coyni?

I configured Selenium Grid and connected it with BrowserStack so the
same tests could run on multiple browsers and environments at the same
time. Jenkins triggered these runs. Together this helped us reach 85%+
automation coverage and cut manual regression effort by 75%.

### Q21. How have you done API testing?

I use REST Assured with Java. I validate status codes, response body
fields, headers and response time for GET, POST, PUT and DELETE calls. I
also check that data returned by the API matches what the UI shows and
what is stored in the database. For example, after placing a
transaction, I verify the API response and then confirm the record with
SQL.

### Q22. What SQL do you use for validation?

Mostly `SELECT` queries with `WHERE`, `JOIN`, `GROUP BY`, `ORDER BY` and
`COUNT`. I use them to confirm that a UI action or API call actually
saved the right data, for example checking a user's balance, a
transaction status or a registration record. I also use SQL to check
test data before and after a run.

### Q23. How do you set up CI/CD for automation?

On Fanatics, I integrated the Playwright regression suites into GitHub
Actions. The workflow installs dependencies, runs tests in parallel, and
publishes reports, so every build gets continuous validation. On Coyni,
I used Jenkins to trigger Selenium Grid and BrowserStack runs. The goal
is quick feedback on every change without anyone running tests manually.

### Q24. What do you do when a test is flaky?

First I find the cause instead of just rerunning: timing issues,
unstable locators, shared test data, environment slowness, or order
dependency. Then I fix it at the root: better locators, proper waits or
web-first assertions, independent test data, and clean preconditions.
Retries can be a safety net in CI, but I do not use them to hide real
problems. In Playwright, traces help a lot to see what actually
happened.

### Q25. How do you use Java OOP concepts in your framework?

Encapsulation in page objects, where locators are private and actions
are public methods. Inheritance in a BaseTest and BasePage class for
common setup and utilities. Abstraction through interfaces or helper
layers so tests do not depend on implementation. Polymorphism in cases
like handling different browsers or drivers through a common driver
reference.

### Q26. Why TypeScript for Playwright, and what is async/await?

TypeScript gives type safety, better auto-complete and early error
detection, which matters in a large framework. Browser actions are
asynchronous, so we use async/await to make steps run in order.
Forgetting an `await` is a common cause of flaky behaviour, so I always
check for it during code review.

### Q27. Which test management tools have you used and how?

TestRail and Zephyr Scale for test cases, test runs and reports, and
Jira for stories and defects. I link test cases to requirements for
traceability and update execution status after each cycle. With the AI
workflow, some of the test-case drafting and defect summarisation became
faster while still being reviewed by me.

## Level 3: Advanced

### Q28. Walk me through the architecture of the Casino framework you designed.

It is a Playwright and TypeScript framework with a layered design. The
test layer holds scenarios like game launch, bet placement, insufficient
funds, favorites and search. Below that is a page and component layer
for the lobby, filters, categories, game tiles and promotions. Then a
utility layer for game discovery, navigation and game-state validation,
and finally config and data. Because of this separation, a UI change
usually affects only one component, not every test.

### Q29. What reusable components did you build and why?

Components for game discovery, lobby navigation, filters, categories,
favorites, game launch, betting flows and game-state validation. Casino
has hundreds of games and many repeated actions, so reusable components
keep tests short and consistent and make new tests quick to add.

### Q30. How do you manage preconditions and dependencies in tests?

I keep them in dedicated modules or fixtures, for example: a logged-in
user, a user with enough balance, or a user with no balance for the
insufficient funds scenario. Tests call these instead of repeating
steps, which keeps them independent and stable. I did this in Coyni with
a dependency and precondition module, and on Fanatics through standards
defined in the spec files.

### Q31. How do you validate game state in a casino test?

I check what the user sees and what the system confirms. After launching
a game, I verify that it loads in the right container, the game is
playable, and the balance and bet amount are correct after a bet. For
game results, I validate the outcome shown and the resulting balance. I
keep this logic in reusable validation helpers.

### Q32. Explain your AI-assisted QA workflow in simple terms.

I use VS Code with GitHub Copilot, Claude Code and Kiro as assistants. I
give them the requirement, they help analyse it and draft test
scenarios, then generate Playwright scripts that follow our framework
standards. They also help prepare preconditions, summarise defects and
maintain regression tests. I always review and run what they generate.
The AI speeds up the repetitive work, and I keep ownership of quality.

### Q33. What is Playwright MCP and how did you use it?

MCP, Model Context Protocol, lets an AI assistant connect to tools. With
Playwright MCP, the assistant can interact with a real browser, so it
can see the page and write more accurate scripts and locators instead of
guessing. I used it along with Claude Code and Copilot to speed up
scripting. I also have exposure to Jira MCP and Git MCP for working with
tickets and repository tasks through the same kind of workflow.

### Q34. What is Spec-Driven Development, and what are AGENTS.md and steering files?

In spec-driven development, we first write clear specs and rules, and
then the AI and the team work against them. AGENTS.md and steering files
are where I document the framework standards: folder structure, naming,
locator rules, how to handle preconditions, and what the AI should or
should not do. Because of this, AI-generated code looks the same as
hand-written code, the framework stays consistent, and
requirement-to-test traceability improves.

### Q35. How do you make sure AI-generated tests are reliable?

Three things: clear rules, review, and execution. The steering files
give the AI boundaries. Then I review every generated script like I
would review a teammate's pull request, checking locators, assertions
and test data. And nothing is accepted until it runs successfully and
fails correctly when it should. AI can make mistakes or invent steps, so
human review is not optional.

### Q36. You mention 95% less manual test-case creation and 80% less scripting effort. How did you arrive at those numbers?

Walk through it honestly: I compared the time we normally spent on a set
of tasks before and after the AI workflow. For test cases, drafting from
requirements used to be fully manual, and with the AI plus spec files
the first draft came out in minutes, with me reviewing and refining. For
scripting, the AI generated the base script and I adjusted it. So the
percentages show effort reduced for the drafting and scripting work, not
the total QA effort. Before the interview, decide the exact baseline you
will quote, such as hours per feature, so your answer sounds precise.

### Q37. What are the risks of using AI in QA and how do you manage them?

The main risks are wrong or made-up output, inconsistent code, and
sharing sensitive information. I handle them by using spec and steering
files for consistency, reviewing and running everything, and not putting
confidential data into prompts beyond what our company allows. I treat
AI as an assistant that speeds me up, not as a replacement for
judgement.

### Q38. You contributed to migrating legacy test assets to JavaScript-based frameworks on Wish. How would you approach a migration like that?

I start by listing existing tests and picking the most valuable ones
first, usually the critical flows and the stable regression tests. Then
I set up the new framework structure, rebuild page objects, and migrate
in batches while running old and new suites side by side to compare
results. I drop outdated or duplicate tests rather than blindly copying
them. The result was better automation quality and faster releases.

### Q39. How do you keep a framework maintainable as it grows?

Layered architecture, reusable components, clear naming, a single place
for each locator, code reviews, and documented standards. I also review
the suite regularly to remove duplicates and fix flaky tests early. The
standards I put in AGENTS.md and steering files help new team members
and AI tools follow the same patterns.

## Scenario-Based Technical Questions

### S1. A test passes locally but fails in GitHub Actions. What do you do?

I check the CI logs, screenshots, video and trace first. Common causes
are environment differences, slower response in CI, different viewport
or browser version, missing test data, or tests depending on each other.
I reproduce it with the same settings if possible, fix the root cause,
for example a wait condition or data setup, and then rerun multiple
times to confirm it is stable.

### S2. A bet placement test fails once in every ten runs. How do you debug it?

I run it repeatedly with trace on and compare a failed run with a passed
one. I look for timing issues in the game load, a locator that changes,
or test data such as balance being used by another test. After finding
the cause, I fix it, for example by waiting for the game ready state or
using a dedicated user, and then run it in a loop to prove it is stable.

### S3. The release is in two days and you have a large regression suite. What do you run?

I prioritise by risk. First the smoke suite on the new build, then the
critical flows: login, game launch, bet placement, payments or
transactions. Then the areas touched by this release. I run in parallel
in CI to save time, and the remaining low-risk tests run after if time
allows. I tell the team clearly what was covered and what was not.

### S4. A UI change breaks fifty tests. What went wrong and how do you prevent it?

Usually it means locators or flows were repeated across tests instead of
living in one place. I fix the page object or component once and the
tests recover. To prevent it, I keep locators centralised, prefer stable
attributes, and talk to developers early about upcoming UI changes.

### S5. The game launches inside an iframe or a new tab. How do you automate it?

In Playwright I use `frameLocator` for iframes and wait for the new page
event for new tabs, then continue on that page object. I wrap this in a
reusable game-launch component so every test handles it the same way.

### S6. How do you automate the insufficient funds scenario?

I use a precondition that prepares a user with a low or zero balance, so
the test does not depend on previous runs. Then the test tries to place
a bet and verifies the correct error message and that the balance and
bet state are unchanged. Keeping data independent makes it reliable.

### S7. How would you verify that a UI action really saved the correct data?

I check three layers: the UI shows the right result, the API response is
correct, and the database record matches, using a SQL query. If all
three agree, I am confident the feature works end to end.

## Quick Practice: Short Coding and SQL Questions

Interviewers often ask a few small questions from the languages you
listed. Practise these:

-   Reverse a string in Java or TypeScript, and check for a palindrome.

-   Count the occurrences of each character in a string using a map.

-   Find duplicates in an array and remove them.

-   **Second highest salary in SQL:**

    ``` sql
    SELECT MAX(salary)
    FROM employee
    WHERE salary < (SELECT MAX(salary) FROM employee);
    ```

-   **Find duplicate records in SQL:**

    ``` sql
    SELECT email, COUNT(*)
    FROM users
    GROUP BY email
    HAVING COUNT(*) > 1;
    ```

-   **Difference between INNER JOIN and LEFT JOIN:** INNER JOIN returns
    only matching rows; LEFT JOIN returns all rows from the left table
    plus matching rows from the right.

-   Difference between `==` and `equals()` in Java, and `let` vs `const`
    vs `var` in JavaScript.

## Level 1: Basics

### M1. What are your roles and responsibilities in your current project?

On Fanatics, I design and maintain the Playwright and TypeScript
automation framework for the Casino application. I automate scenarios
like game launch, bet placement, insufficient funds, favorites, search,
categories and navigation. I integrate the regression suites into GitHub
Actions, analyse failures, log defects, and I also use AI tools to speed
up test generation and maintenance. In short, I own the automation side
from requirement analysis to CI execution.

### M2. What does a typical sprint look like for you?

At the start, I review the stories and acceptance criteria and clarify
doubts with the product owner or developers. During the sprint, I
prepare test scenarios, automate the new features, and update the
regression suite. Near the end, I run the full regression in CI, report
results, and retest fixes. I also join the daily stand-up, sprint review
and retrospective.

### M3. How do you work with developers and product owners?

I talk early and often. I ask questions when a requirement is unclear
instead of assuming, I share defects with clear steps and evidence, and
I ask for stable locators or test IDs when the UI makes automation hard.
I treat developers as partners, so the conversation is about fixing the
product, not about blame.

### M4. How do you prioritise your testing when time is short?

I go by risk and business impact. Critical flows such as login, game
launch, bet placement and payments come first. Then areas touched by the
change, and then lower-risk areas. I keep the team informed about what
is covered and what is not, so decisions are made with full information.

### M5. How do you report test status?

Through clear numbers and a short summary: tests executed, passed,
failed, blocked, defects open by severity, and any risk to the release.
Our CI generates automated reports after each run, so I add context,
like which failures are real defects and which are environment or data
issues.

## Level 2: Intermediate

### M6. Tell me about something you owned end to end.

The Casino automation framework on Fanatics. I designed the structure,
built reusable components for game discovery, lobby navigation, filters,
favorites, game launch and betting flows, and defined the standards for
the team through AGENTS.md and steering files. Then I connected the
regression suites to GitHub Actions with parallel execution and
reporting. It gave the team a consistent and scalable base to keep
adding tests.

### M7. How do you write a good defect report?

A clear title, environment and build details, exact steps to reproduce,
expected versus actual result, severity, and evidence such as
screenshots, video, logs or trace. I make it so a developer can
reproduce it without asking me anything. I also use AI to help summarise
defects, then I check the content before logging.

### M8. A developer says it is not a bug. What do you do?

I stay calm and go back to the requirement or acceptance criteria. I
show the evidence and compare the expected behaviour with what happened.
If the requirement itself is unclear, I involve the product owner or BA
to decide. The goal is the right outcome for the product, not winning
the argument.

### M9. You find a critical defect right before release. What do you do?

I confirm it first: reproduce it, check the build and environment, and
collect evidence. Then I inform the manager, the developers and the
product owner immediately with the impact. I let the business decide
whether to fix now, delay the release or accept the risk, and I offer to
retest quickly once a fix is ready.

### M10. Tell me about a challenge you faced and how you solved it.

On Coyni, we had a lot of repeated code and unstable tests because every
test handled its own setup. I built reusable keyword libraries and a
dependency and precondition module. That cut code duplication by more
than 50% and made tests much more stable during sprints. You can add one
real example of a specific failure you fixed to make this more personal.

### M11. How do you make sure you have enough test coverage?

I map requirements to test cases through an RTM, include positive,
negative and edge cases, and track which are automated. On Coyni, we
reached 85%+ automation coverage. On Fanatics, spec-driven development
improved requirement-to-test traceability, so gaps are easier to spot.

### M12. How do you handle changing requirements in Agile?

I stay close to the team so I hear about changes early. Because my
framework is modular, a change usually means updating one component or
page object instead of many tests. I update affected test cases,
re-prioritise if needed, and make sure the regression suite reflects the
latest behaviour.

### M13. How have you improved a process in your team?

The biggest one is the AI-assisted workflow. Using GitHub Copilot,
Claude Code, Kiro and Playwright MCP with spec-driven standards, we
reduced manual test-case creation by 95% and automation scripting effort
by 80%. Another one is moving regression into GitHub Actions with
parallel execution so feedback arrives in each build, not at the end of
the cycle.

### M14. How do you help teammates follow the same standards?

I write the standards down where everyone, including the AI tools, can
see them: AGENTS.md and steering files for structure, naming, locator
rules and preconditions. This keeps the framework consistent and makes
it easy for new people to start contributing. I also share knowledge in
reviews and discussions. If you have mentored someone or run a
walkthrough session, mention it here.

## Level 3: Advanced

### M15. How do you decide what to automate and how do you show the value?

I look at how often a test runs, how critical the flow is, how stable
the feature is and how costly it is to run manually. Then I show value
with numbers: on Coyni, manual regression effort fell by 75% and
automation coverage went above 85%. I also talk about faster feedback
and fewer repeated manual cycles.

### M16. How would you convince a sceptical team or manager to adopt AI in QA?

I would start small with a pilot on one feature and measure it: time to
write test cases, time to script, and quality of the output. I would
show that rules in steering files and human review keep the output safe
and consistent. When people see actual time saved and no drop in
quality, adoption gets easy. I would also be open about the limits of AI
so nobody feels it is being oversold.

### M17. Which quality metrics do you track?

Automation coverage, pass rate and flaky test rate, defect counts by
severity, defect leakage to production, execution time of the regression
suite, and requirement-to-test traceability. I use them to find where to
improve instead of just reporting numbers.

### M18. A defect escaped to production. What do you do?

First help fix and verify it quickly. Then do a blameless root cause
analysis: why did our tests miss it? Was there no test, a weak
assertion, bad data or an unclear requirement? I add or improve tests,
update the regression suite, and share the learning with the team so it
does not repeat.

### M19. If you joined a new project as a senior automation engineer, how would you start?

First 30 days: understand the product, users, release process, existing
tests and pain points. Next 30 days: fix the biggest problems, such as
flaky tests or missing CI, and automate the highest-risk flows with a
clean, layered structure. After that: set standards, documentation and
reporting, expand coverage in a planned way, and review progress with
the manager using clear metrics.

### M20. Tell me about a mistake you made and what you learned.

Pick a real but small example. A safe structure: I automated a flow
without confirming one requirement detail, so the test passed while
checking the wrong thing. Once found, I corrected the test and now I
always confirm acceptance criteria and review assertions before
finalising. The learning is to clarify first and review my own work as
critically as I review others.

### M21. How do you handle disagreement with your manager or team lead?

I listen to their reasoning, share mine with facts, and suggest options.
If we still disagree, I respect the decision once it is made and support
it, as long as it is not unethical or unsafe. Most of the time, a
data-based discussion settles it.

## Scenario-Based Managerial Questions

### MS1. The sprint is ending and your automation for a new feature is not ready. What do you do?

I tell the team early, not on the last day. I test the feature manually
to protect the release, automate the most critical scenarios first, and
plan the rest for the next sprint with the manager's agreement. The
release is not blocked, and the automation debt is visible and tracked.

### MS2. Your manager asks you to cut testing to meet a deadline. How do you respond?

I explain the risk with specifics: which areas would be untested and
what could go wrong. Then I propose options: a risk-based subset,
running only critical automated suites in parallel, or releasing with
known risks documented. The manager makes an informed decision, and I
support it.

### MS3. Two releases need your attention at the same time.

I ask for the priorities from the manager and product owner, share my
capacity honestly, and split the work: critical checks first for both,
using automation to cover repeat work. I communicate any risk quickly.

### MS4. The test environment is down and the team is waiting for results.

I raise it with the right team immediately, share the impact, and use
the time productively: writing or refactoring tests, reviewing
scenarios, or checking in a lower environment if allowed. Once the
environment is back, I run smoke tests first to verify it is healthy
before the full suite.

### MS5. A script generated by AI passed review but later gave a wrong result. What do you do?

I would fix the script, find out why my review missed it, for example a
weak assertion, and update the steering files with a rule to prevent the
same mistake. It also reminds the team that AI output needs human review
and a run that fails when it should.

### MS6. Management wants to reduce regression time from many hours to much less. What is your plan?

Parallel execution in CI, removing duplicate and low-value tests, fixing
slow and flaky tests, and splitting the suite into smoke, critical and
full regression so the right level runs for the right trigger. That is
the same approach that helped cut manual regression effort by 75% on
Coyni and gave continuous validation on Fanatics.

# Round 3: HR Round

**Tip:** Keep HR answers short (under two minutes), positive and honest.
Never speak badly about your current company or team. Fill in the
