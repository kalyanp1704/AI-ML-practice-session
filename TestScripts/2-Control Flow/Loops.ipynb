# Python Loops – Definitions and Concepts

## 1. Introduction to Loops in Python

**Definition:** A loop is a programming construct used to execute a block of code repeatedly, either while a condition remains true or until all items in a sequence have been processed.

**Why do we use loops?**

Loops help us avoid writing the same code multiple times. For example, instead of writing five separate `print()` statements to display "Hello", we can use a loop.

Python has two main types of loops:
- `for` loop
- `while` loop

## 2. For Loop

**Definition:** A `for` loop is used to iterate over a sequence of items, such as a range of numbers, a string, a list, or a tuple.

### A. Iterating Over a Range

The `range()` function generates a sequence of integers that can be processed one at a time using a `for` loop.

### B. Iterating Over a String

**Definition:** Iterating over a string means accessing and processing each character in the string one at a time using a loop.

## 3. While Loop

**Definition:** A `while` loop repeatedly/continuesly executes a block of code as long as the specified condition evaluates to `True`.

## 4. Loop Control Statements

**Definition:** Loop control statements modify the normal flow of loop execution. The three commonly used statements are `break`, `continue`, and `pass`.

### A. `break` – Terminate the Loop

**Definition:** The `break` statement immediately terminates the nearest enclosing loop and transfers control to the statement following that loop.

### B. `continue` – Skip the Current Iteration

**Definition:** The `continue` statement skips the remaining statements in the current iteration and proceeds to the next iteration of the loop.

### C. `pass` – Placeholder Statement

**Definition:** The `pass` statement performs no action. It is used as a placeholder when a statement is syntactically required but the code has not yet been implemented.

## 5. Nested Loops

**Definition:** A nested loop is a loop placed inside another loop. The inner loop executes as part of each iteration of the outer loop.

```python
range(5)

# for loop
for i in range(1, 10, 2):
    print(i)
```

```python
## Strings

str = "Kalyan"

for i in str:
    print(i)
```

```python
## While Loop

count = 0

while count <5:
    print(count)
    count = count +1
```

```python
# Loop control statements

# Break Statement

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# Continue 

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
## Pass

for i in range(5):
    if i == 3:
        pass
    print(i)
```

```python
## Nested loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples
# Caliculate the sum of first N natural numbers by using while loop

n = 5
sum = 0
count = 1

while count <= n:
    sum = sum + count
    count = count + 1

print('Sum of First 5 natural numbers:', sum)
```

```python
# using for loop

n = 5
sum = 0
for i in range(11):
    sum = sum + i
print(sum)
```

```python
## Ex - prime number

for num in range(1, 20):
    if num > 1:
        for i in range(2,num):
            if num % i == 0:
                break
        else:
            print(num)
```
