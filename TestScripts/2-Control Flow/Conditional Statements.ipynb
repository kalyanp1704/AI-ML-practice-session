```python
#n If Statement - if statement evaluates a condition and executes the block of code within it.

age = 20

if age>=18:
    print("TRUE")
    
```

```python
# ELSE = else statement executes a block of code if the condition is false in the If statement.

age = 16

if age >= 18:
    print("Your are eligible for voting")
else:
    print("Your are minor")
```

```python
#Elif - elif statement allows to check multiple conditions. it statnds for Else if

age = 17

if age < 13:
    print("VYour are child")
elif age < 18:
    print("Your are teenagger")
else:
    print("Your are Adult")
```

```python
## Nested condition statement

## we can place one or more if, elif, or else statements inside another if, elif or else statement to created nested conditional statements.

# number even, odd, negative

num = int(input("Enter the number"))
if num > 0:
    print("Positive")
    if num %2 == 0:
        print("Even")
    else:
        print("Odd")
else:   
    print("Number is Zero or Negative")
```

```python
#Examples
# Determine if a year is leap year using nested condition statement

year = int(input("Enter the year"))

if year%4 ==0:
    if year%100 == 0:
        if year%400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")    
```
