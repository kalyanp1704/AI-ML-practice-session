#Comparison Operators - A comparison operator is used to compare two values.

| Operator | Meaning                  | Example    | Result  |
| -------- | ------------------------ | ---------- | ------- |
| `==`     | Equal to                 | `10 == 10` | `True`  |
| `!=`     | Not equal to             | `10 != 5`  | `True`  |
| `>`      | Greater than             | `10 > 5`   | `True`  |
| `<`      | Less than                | `10 < 5`   | `False` |
| `>=`     | Greater than or equal to | `10 >= 10` | `True`  |
| `<=`     | Less than or equal to    | `5 <= 10`  | `True`  |


Super easy memory trick:

== → Same / != → Different / < → Smaller / <= → Smaller or Same / > → Bigger / >= → Bigger or Same 


#Logical Operator - Logical operators are used to combine multiple conditions.

| Operator | Meaning                             | Easy memory  |
| -------- | ----------------------------------- | ------------ |
| `and`    | Both conditions must be True        | **Both**     |
| `or`     | At least one condition must be True | **Any one**  |
| `not`    | Reverses True/False                 | **Opposite** |
