```python
# Declaring and Assigning variables

age = 32
height = 5.9
name = "Kalyan"
is_student = True

# Printing the Variables
print("Age : ", age)
print("Height : ", height)
print("Name:", name)
print("Is_Student :", is_student)
```

```python
## understanding variable types
## Python is dynamically typed, type of variable is detrmined at runtime

age = 32
height = 5.9
name = "Kalyan"
is_student = True

print(type(name))
```

```python
## Type checking and conversion

type(age)

age = 25 # Number (int)
print(type(age))

#Type conversion
age_str = str(age)   #Convert number to text
print(age_str)       #stores the converted value
print(type(age_str)) #tells us the data type

height = 5.9
type(height)
int(height) #convert float -> int
```

Dynamic Typing --important

Dynamic typing means Python automatically determines the data type of a variable when you assign a value.

No need to specify int, string, float, etc.

```python
var = 10
print(var, type(var))

var = "Kalyan"
print(var, type(var))
```

```python
#input() function is used to take input from user. 
# It always returns the input as string and 
# we can convert it to other data types using type conversion function.

age = int(input("Enter my age:"))
print(age, type(age))
```
